## import packages

````python
import scanpy as sc
import anndata
from SpaMOMA.SpaMOMA import SpaMOMA
````

## Specify Slice Metadata

````python
save_directory = "./SpaMOMA_output/MH/"  
spamoma = SpaMOMA(save_dir=save_directory)
for sample_id in ['S1','S2','S3','S4','S5','S6','S7','S8']:
    spamoma.add_slice(
        sample_id=sample_id,  
        raw_data_path=f"./datasets/MH/{sample_id}.h5ad",  
        feature="rna"  
    )

spamoma.set_reference('S1')
print(f"aligning {[sid for sid in spamoma.input_slice_dict.keys() if sid != spamoma.reference_sample_id]} to {spamoma.reference_sample_id}")
````

## Data Preprocessing

````python
spamoma.preprocess_input_slices()
````

## Generate Spatial Pattern Images

````python
spamoma.generate_imgs()
````

## Model Training

````python
spamoma.train(coarse_epochs=1000,fine_epochs=1000,parallel_mode=True,coarse_gpu=1,fine_gpu=1)
````

## Slice Alignment

````python
spamoma.align_slices()
````

## Visualize Alignment Results

````python
sample_id_list=['S1','S2','S3','S4','S5','S6','S7','S8']
adata_list=[]
for sample_id in sample_id_list:
    adata=spamoma.read_aligned_data(sample_id)
    adata_list.append(adata)
    
adata_concat=anndata.concat(adata_list)
sc.pl.spatial(adata_concat,color='ground_truth',spot_size=20,basis='aligned_spatial')
````
